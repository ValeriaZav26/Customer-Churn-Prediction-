## 🌐 Web Scraping: Book Data Collection

In this notebook we scrape data from:

http://books.toscrape.com/

This site is a **demo online bookstore** built specifically for practicing **web scraping**. It contains information on roughly **1000 books**, including their **titles, prices, and ratings**.

**Web scraping** is a technique for **automatically extracting data from websites** and turning it into a structured format (e.g. a table or dataset) for further analysis.

Using the Python libraries **Requests** and **BeautifulSoup**, we will:

- send HTTP requests to the site
- parse the HTML structure of each page
- extract the required information about every book
- store the collected data for further analysis

Since the site has multiple pages, the script walks through **all pages one by one** to collect the full dataset (the number of pages is detected automatically instead of being hardcoded).

The last part of the notebook contains an **extended exploratory data analysis (EDA)** of the collected data.

```python
import re

import requests
from bs4 import BeautifulSoup
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Quick look at a single page

Before scraping the whole site, let's fetch one page and inspect its structure.

```python
url = "http://books.toscrape.com/"

response = requests.get(url)  # send a request to the site and get the page HTML

soup = BeautifulSoup(response.text, "html.parser")  # turn the HTML into a structure we can search

books = soup.find_all("article", class_="product_pod")  # find all book cards on the page

len(books)
```

### Extracting fields from one page

Let's test extracting all the fields we need (title, price, rating, link) from a single page before scaling this up to the whole site.

```python
data = []

for book in books:
    title = book.find("h3").find("a")["title"]  # extract the book title from the HTML structure

    price = book.find("p", class_="price_color").text
    price = float(price.replace("£", "").replace("Â", ""))  # strip the currency symbol and convert to a number

    rating_class = book.find("p", class_="star-rating")["class"][1]  # rating is stored as an HTML class (e.g. Three, Four)

    link = book.find("a")["href"]  # relative link to the book's own page

    data.append({
        "title": title,
        "price": price,
        "rating": rating_class,
        "link": link
    })  # add the book's info to the list

df = pd.DataFrame(data)
df.head()
```

### Scraping the whole site

Instead of hardcoding the number of pages, the loop stops automatically once it hits a page with no book cards (or a 404) — this way the scraper keeps working even if the site adds or removes pages later.

We also fixed a small bug from the original version: the `link` field was being extracted on the test page above but was missing from the final collection loop.

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
                  "(KHTML, like Gecko) Chrome/124.0.0.0 Safari/537.36"
}  # a normal browser User-Agent instead of the default python-requests one

all_data = []
page = 1

while True:  # walk through all pages until there are none left, instead of a hardcoded range(1, 51)

    if page == 1:
        url = "http://books.toscrape.com/"
    else:
        url = f"http://books.toscrape.com/catalogue/page-{page}.html"  # build the page URL automatically

    response = requests.get(url, headers=headers)

    if response.status_code == 404:
        break  # no more pages

    response.raise_for_status()  # check the request succeeded
    response.encoding = "utf-8"

    soup = BeautifulSoup(response.text, "html.parser")
    books = soup.find_all("article", class_="product_pod")

    if not books:
        break  # safety net in case a page loads but has no book cards

    for book in books:

        # Title
        title = book.find("h3").find("a")["title"]

        # Price
        price_text = book.find("p", class_="price_color").text
        price = float(re.search(r"\d+\.\d+", price_text).group())

        # Rating
        rating = book.find("p", class_="star-rating")["class"][1]

        # Link (fixed: this field was missing from the original collection loop)
        link = book.find("a")["href"]

        all_data.append({
            "title": title,
            "price": price,
            "rating": rating,
            "link": link
        })

    page += 1

print(f"Pages scraped: {page - 1}")
print(f"Books collected: {len(all_data)}")

df = pd.DataFrame(all_data)
df
```

```python
# convert the text rating into a numeric one

rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

df["rating"] = df["rating"].map(rating_map)
df.head()
```

```python
# drop exact duplicate books, just in case a page was scraped twice

before = len(df)
df = df.drop_duplicates(subset=["link"]).reset_index(drop=True)
print(f"Removed {before - len(df)} duplicate rows, {len(df)} rows remain")
```

## 📊 Exploratory Data Analysis (EDA)



### 1. Dataset overview

```python
print(f"Dataset size: {df.shape[0]} rows, {df.shape[1]} columns")
df.info()
```

```python
print(df.isna().sum())
print(df.duplicated(subset=["link"]).sum())
```

```python
df.describe()
```

### 2. Rating distribution

```python
rating_counts = df["rating"].value_counts().sort_index()
rating_pct = (rating_counts / len(df) * 100).round(1)

plt.figure(figsize=(7, 4))
sns.barplot(x=rating_counts.index, y=rating_counts.values, palette="viridis")
plt.title("Number of books by rating")
plt.xlabel("Rating (stars)")
plt.ylabel("Number of books")
plt.show()

pd.DataFrame({"count": rating_counts, "percent": rating_pct})
```

### 3. Price distribution

```python
print("Average price:", round(df["price"].mean(), 2))
print("Median price:", round(df["price"].median(), 2))
print("Most expensive book:", df.loc[df["price"].idxmax(), "title"])
print("Cheapest book:", df.loc[df["price"].idxmin(), "title"])

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

sns.histplot(df["price"], bins=30, kde=True, ax=axes[0])
axes[0].set_title("Price histogram")
axes[0].set_xlabel("Price, £")

sns.boxplot(x=df["price"], ax=axes[1])
axes[1].set_title("Price boxplot (outlier check)")
axes[1].set_xlabel("Price, £")

plt.tight_layout()
plt.show()

print(f"Skewness: {df['price'].skew():.3f}  (0 = symmetric distribution)")
```

### 4. Price by rating

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

df.groupby("rating")["price"].mean().plot(kind="bar", ax=axes[0], color="teal")
axes[0].set_title("Average price by rating")
axes[0].set_xlabel("Rating (stars)")
axes[0].set_ylabel("Average price, £")

sns.boxplot(data=df, x="rating", y="price", palette="viridis", ax=axes[1])
axes[1].set_title("Price distribution by rating")
axes[1].set_xlabel("Rating (stars)")
axes[1].set_ylabel("Price, £")

plt.tight_layout()
plt.show()

df.groupby("rating")["price"].agg(["mean", "median", "std", "min", "max", "count"])
```

### 5. Correlation between price and rating

Both Pearson (linear relationship) and Spearman (monotonic, rank-based) correlation are reported.

```python
pearson_corr = df["rating"].corr(df["price"], method="pearson")
spearman_corr = df["rating"].corr(df["price"], method="spearman")

print(f"Pearson correlation (rating vs price):  {pearson_corr:.3f}")
print(f"Spearman correlation (rating vs price): {spearman_corr:.3f}")

corr_matrix = df[["price", "rating"]].corr()

plt.figure(figsize=(4, 3.5))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", vmin=-1, vmax=1, fmt=".3f")
plt.title("Correlation heatmap")
plt.tight_layout()
plt.show()
```

### 6. Most expensive / cheapest books

```python
print("Top 10 most expensive books:")
display(df.nlargest(10, "price")[["title", "price", "rating"]])

print("\nTop 10 cheapest books:")
display(df.nsmallest(10, "price")[["title", "price", "rating"]])
```

### 7. Title length analysis

```python
df["title_char_len"] = df["title"].str.len()
df["title_word_count"] = df["title"].str.split().str.len()

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
sns.histplot(df["title_char_len"], bins=25, kde=True, ax=axes[0])
axes[0].set_title("Title length (characters)")
axes[0].set_xlabel("Number of characters")

sns.histplot(df["title_word_count"], bins=range(1, df["title_word_count"].max() + 2), ax=axes[1])
axes[1].set_title("Title length (words)")
axes[1].set_xlabel("Number of words")

plt.tight_layout()
plt.show()

len_price_corr = df["title_char_len"].corr(df["price"])
print(f"Correlation between title length (characters) and price: {len_price_corr:.3f}")
```

### 8. Most frequent words in titles

```python
from collections import Counter

stopwords = {
    "the", "a", "an", "of", "and", "in", "to", "on", "for", "with", "at",
    "by", "from", "is", "it", "this", "that", "as", "or", "be", "are",
}

words = (
    df["title"]
    .str.lower()
    .str.replace(r"[^a-z0-9\s]", " ", regex=True)
    .str.split()
    .explode()
)
words = words[~words.isin(stopwords) & (words.str.len() > 2)]

top_words = Counter(words).most_common(20)
top_words_df = pd.DataFrame(top_words, columns=["word", "count"])

plt.figure(figsize=(8, 6))
sns.barplot(data=top_words_df, x="count", y="word", palette="viridis")
plt.title("Top 20 most frequent words in book titles")
plt.xlabel("Frequency")
plt.ylabel("Word")
plt.tight_layout()
plt.show()

top_words_df
```

### 9. Price segmentation

```python
bins = [0, 20, 35, 50, float("inf")]
labels = ["Budget (<£20)", "Mid-range (£20-35)", "Premium (£35-50)", "Luxury (>£50)"]
df["price_segment"] = pd.cut(df["price"], bins=bins, labels=labels)

segment_counts = df["price_segment"].value_counts().reindex(labels)

plt.figure(figsize=(7, 4))
sns.barplot(x=segment_counts.index, y=segment_counts.values, palette="magma")
plt.title("Books by price segment")
plt.xlabel("Price segment")
plt.ylabel("Number of books")
plt.xticks(rotation=15)
plt.show()

segment_counts.to_frame("count")
```

## Key takeaways

- The rating distribution is fairly even across the 1–5 scale.
- The average book price is around **£35**.
- **Book rating has no meaningful effect on price** — both the Pearson/Spearman correlations and the boxplots by rating group show essentially no relationship, which makes sense since ratings on this demo site are assigned independently of price.
- Price is right-skewed with a moderate spread; the IQR method flags a small number of higher-priced books as statistical outliers rather than data errors.
- Title length and word choice show no strong relationship with price, but the word-frequency analysis is a useful demonstration of basic text EDA on scraped catalog data.
